```python
import cv2
import tkinter as tk
from ultralytics import YOLO

MODEL_PATH = "/Users/artemsaltanovskyi/Documents/2025-2026-2/2MCTE2-Advancded_AI/exam/AIProject2026/Task02/Task02_PreLabeledDataset/Results/Models/YOLO_Results/yolo11n_test_run/weights/best.pt"
VIDEO_PATH = "/Users/artemsaltanovskyi/Documents/2025-2026-2/2MCTE2-Advancded_AI/exam/AIProject2026/Data/Dataset_Prelabeled/Video/road.MP4"

model = YOLO(MODEL_PATH)

cap = cv2.VideoCapture(VIDEO_PATH)

if not cap.isOpened():
    raise FileNotFoundError(f"Unable to open the video: {VIDEO_PATH}")

root = tk.Tk()
screen_w = root.winfo_screenwidth()
screen_h = root.winfo_screenheight()
root.destroy()

max_w = int(screen_w * 0.9)
max_h = int(screen_h * 0.85)

window_name = "YOLO Video Detection"
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)

def resize_with_aspect_ratio(image, max_width, max_height):
    h, w = image.shape[:2]
    scale = min(max_width / w, max_height / h)
    new_w = int(w * scale)
    new_h = int(h * scale)
    return cv2.resize(image, (new_w, new_h), interpolation=cv2.INTER_AREA)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model.predict(
        source=frame,
        imgsz=640,
        conf=0.25,
        verbose=False
    )

    annotated_frame = results[0].plot()
    display_frame = resize_with_aspect_ratio(annotated_frame, max_w, max_h)

    cv2.imshow(window_name, display_frame)

    key = cv2.waitKey(1) & 0xFF
    if key == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
