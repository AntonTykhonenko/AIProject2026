```python
import os
import cv2
import pandas as pd

VIDEO_PATH = "/Users/artemsaltanovskyi/Documents/2025-2026-2/2MCTE2-Advancded_AI/exam/AIProject2026/Data/Video/road.mp4"
CSV_PATH = "/Users/artemsaltanovskyi/Documents/2025-2026-2/2MCTE2-Advancded_AI/exam/AIProject2026/Data/Csv/G.csv"
OUTPUT_DIR = "/Users/artemsaltanovskyi/Documents/2025-2026-2/2MCTE2-Advancded_AI/exam/AIProject2026/Data/FramesFromCsv"
FRAME_OFFSET = 0

os.makedirs(OUTPUT_DIR, exist_ok=True)

df = pd.read_csv(CSV_PATH, usecols=["frame_num"])
csv_frames = sorted(df["frame_num"].dropna().astype(int).unique().tolist())

video_to_csv = {}
for csv_frame in csv_frames:
    video_frame = csv_frame + FRAME_OFFSET
    if video_frame >= 0:
        video_to_csv[video_frame] = csv_frame

target_video_frames = sorted(video_to_csv.keys())
target_set = set(target_video_frames)

cap = cv2.VideoCapture(VIDEO_PATH)
if not cap.isOpened():
    raise FileNotFoundError(VIDEO_PATH)

current_frame = 0
saved = 0
target_idx = 0

while cap.isOpened() and target_idx < len(target_video_frames):
    ret, frame = cap.read()
    if not ret:
        break

    target_frame = target_video_frames[target_idx]

    if current_frame == target_frame:
        csv_frame = video_to_csv[current_frame]
        out_path = os.path.join(OUTPUT_DIR, f"frame_{csv_frame:06d}.jpg")
        cv2.imwrite(out_path, frame)
        saved += 1
        target_idx += 1

    current_frame += 1

cap.release()

print(f"Saved: {saved}")
print(OUTPUT_DIR)
```
